```python
import numpy as np
import xarray as xr

import gsw

import os.path as op

from matplotlib import ticker
import matplotlib.colors as clr
import matplotlib.pyplot as plt
plt.rcParams['pcolor.shading'] = 'auto'
%matplotlib inline
```

```python
ddir = "/storage/tartar/takaya/swot/"
```

```python
dsS = xr.open_dataset(op.join(ddir,"SWOT_with_IT/SWOT_L3_LR_SSH_HRET_CALVAL_001_v2.0.1.nc"))
dsH = xr.open_dataset(op.join(ddir,"HYCOM-tidal-correction/hycom_ssh_pass01_swot1day_v2.0.1.nc"))
dsS, dsH
```

```python
dsS.ssha_filt.isel(cycle_num=1).plot(rasterized=True)
plt.show()
dsH.hycom_total.isel(num_cycle=1).plot(rasterized=True)
plt.show()
```

```python
for i in range(1,29):
    dsS = xr.open_dataset(op.join(ddir,"SWOT_with_IT/SWOT_L3_LR_SSH_HRET_CALVAL_%03d_v2.0.1.nc"
                                  % i),
                          chunks={"cycle_num":1}
                         )
    dsH = xr.open_dataset(op.join(ddir,"HYCOM-tidal-correction/hycom_ssh_pass%02d_swot1day_v2.0.1.nc"
                                  % i),
                          chunks={"num_cycle":1}
                         )

    print(dsS.pass_num.values, dsH.num_pass.values)
    ssh_detided = dsS.ssha_filt - dsH.hycom_total.isel(num_pass=0).data

    ssh_detided.isel(cycle_num=i).plot(rasterized=True)
    plt.show()

    dsave = ssh_detided.to_dataset(name="ssha_detided")
    dsave.to_zarr(op.join(ddir,"SWOT_no_IT/CalVal_pass%03d_v2.0.1.zarr"
                          % i), mode="w"
                 )
    dsave.close()
    dsS.close(), dsH.close()
```

```python
for i in range(1,29):
    dsS = xr.open_dataset(op.join(ddir,"with_HRET/SWOT_with_IT/SWOT_L3_LR_SSH_HRET_CALVAL_%03d_v2.0.1.nc"
                                  % i),
                          chunks={"cycle_num":1}
                         )

    # print(dsS.pass_num.values, dsH.num_pass.values)
    ssh_detided = dsS.ssha_filt - dsS.hret

    ssh_detided.isel(cycle_num=i).plot(rasterized=True)
    plt.show()

    dsave = ssh_detided.to_dataset(name="ssha_detided")
    dsave.to_zarr(op.join(ddir,"SWOT_no_IT/HRET/CalVal_pass%03d_v2.0.1.zarr"
                          % i), mode="w"
                 )
    dsave.close()
    dsS.close()
```
